```python
from sklearn.metrics import make_scorer, roc_auc_score
def gini_coefficient(y_true, y_pred):
    """
    Calculate the Gini coefficient using predictions and true labels.
    
    Parameters:
    y_true (array-like): True binary labels.
    y_pred (array-like): Predicted probabilities.
    
    Returns:
    float: Gini coefficient.
    """
    auc = roc_auc_score(y_true, y_pred)  # AUC calculation
    return 2 * auc - 1  # Gini coefficient
```

```python
duplicate_sub = submissions[~submissions['GROUND_TRUTH'].isnull()]
```

```python
duplicate_sub['GROUND_TRUTH'] = duplicate_sub['GROUND_TRUTH'].astype(int)
```

```python
gini_coefficient(duplicate_sub['GROUND_TRUTH'], duplicate_sub['TARGET'])
```

```python
y_true = duplicate_sub['GROUND_TRUTH'].values 
y_pred = duplicate_sub['TARGET'].values

y_true = np.concatenate([y_true, np.zeros(20000), np.ones(5000)])
y_pred = np.concatenate([y_pred, np.zeros(20000), np.ones(5000)])

gini_coefficient(y_true, y_pred)
```
