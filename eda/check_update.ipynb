```python
import pandas as pd
import numpy as np
```

## Prev application

```python
df1 = pd.read_parquet('../data/dseb63_pos_cash_gb.parquet')
df2 = pd.read_parquet('../data/dseb63_pos_cash_gb_v2.parquet')
```

```python
orignal_cols = df1.columns 
new_cols = df2.columns

add_cols = []
del_cols = []

for col in orignal_cols:
    if col not in new_cols:
        del_cols.append(col)

for col in new_cols:
    if col not in orignal_cols:
        add_cols.append(col)

print('New columns:')
for col in add_cols:
    print(f'- {col}')

print('Removed columns:')
for col in del_cols:
    print(f'- {col}')
```
